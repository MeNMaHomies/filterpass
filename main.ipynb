```python
import numpy as np
import soundfile as sf
import IPython.display as ipd
import matplotlib.pyplot as plt
from scripts.data_augmentation import (
    WhiteNoise, PinkNoise, BrownNoise,
    BandpassFilter, MicColoration, CodecDistortion,
    PacketLoss, TimeStretch, RandomGain,
    Compose, OneOf, SomeOf,
    telephony, full, light, voip, pstn, meeting_room, noisy_mobile, adversarial,
    augment_directory, augment_file
)

audio_path = "./sample/gen_0.wav"
audio, sr = sf.read(audio_path, dtype="float32", always_2d=False)

augmentations = {
    "Original": None,
    "WhiteNoise": WhiteNoise(snr_db=(15, 25), p=1.0),
    "PinkNoise": PinkNoise(snr_db=(15, 25), p=1.0),
    "BrownNoise": BrownNoise(snr_db=(15, 20), p=1.0),
    "BandpassFilter": BandpassFilter(low_hz=300, high_hz=3400, p=1.0),
    "MicColoration": MicColoration(freq_hz=1500, gain_db=3.0, p=1.0),
    "CodecDistortion": CodecDistortion(threshold=0.85, p=1.0),
    "PacketLoss (silence)": PacketLoss(rate=0.10, fill="silence", p=1.0),
    "PacketLoss (repeat)": PacketLoss(rate=0.10, fill="repeat", p=1.0),
    "PacketLoss (noise)": PacketLoss(rate=0.10, fill="noise", p=1.0),
    "TimeStretch (slow)": TimeStretch(rate_range=(0.85, 0.85), p=1.0),
    "TimeStretch (fast)": TimeStretch(rate_range=(1.15, 1.15), p=1.0),
    "RandomGain": RandomGain(db_range=(-6, 6), p=1.0),
}

presets = {
    "Preset: telephony": telephony(),
    "Preset: full": full(),
    "Preset: light": light(),
    "Preset: voip": voip(),
    "Preset: pstn": pstn(),
    "Preset: meeting_room": meeting_room(),
    "Preset: noisy_mobile": noisy_mobile(),
    "Preset: adversarial": adversarial(),
}

all_augmentations = {**augmentations, **presets}

# n = len(all_augmentations)
# fig, axes = plt.subplots(n, 1, figsize=(14, 2.5 * n), sharex=True)

# for ax, (name, aug) in zip(axes, all_augmentations.items()):
#     if aug is None:
#         result = audio
#     else:
#         result = aug(audio, sr)

#     librosa.display.waveshow(result, sr=sr, ax=ax, alpha=0.7)
#     ax.set_title(name, fontsize=11, loc="left")
#     ax.set_ylabel("Amplitude")

#     print(f"{name:25s} | len={len(result):>7d} | min={result.min():.4f} | max={result.max():.4f} | rms={np.sqrt(np.mean(result**2)):.4f}")

# plt.tight_layout()
# plt.show()

print("\n--- Audio Playback ---")
for name, aug in all_augmentations.items():
    result = aug(audio, sr) if aug is not None else audio
    print(f"\n{name}:")
    ipd.display(ipd.Audio(result, rate=sr))
```
